```python
customers = spark.table("bronze_customers")
orders = spark.table('bronze_orders')
products = spark.table('bronze_products')
```

```python
from pyspark.sql.functions import trim,col,lower,current_timestamp
silver_customers = (customers
                    .dropDuplicates(['id'])
                    .withColumn('firstName',trim(col('firstName')))
                    .withColumn('lastName',trim(col('lastName')))
                    .withColumn('email',lower(col('email')))
                    .withColumn('gender',lower(col('gender')))
                    .withColumn('ingestion_timestamps',current_timestamp())
                    ) 
```

```python
silver_orders = (orders
                 .dropDuplicates(['id'])
                 .withColumn('total',col('total').cast('double'))
                 .withColumn('discountedTotal',col('discountedTotal').cast('double'))
                 .withColumn('userId',col('userId').cast('int'))
                 .withColumn('totalProducts',col('totalProducts').cast('int'))
                 .withColumn('totalQuantity',col('totalQuantity').cast('int'))
                 .withColumn('ingestion_timestamp',current_timestamp()))
```

```python
silver_products = (products
                   .dropDuplicates(['id'])
                   .withColumn('title',trim(col('title')))
                   .withColumn('category',lower(col('category')))
                   .withColumn('price',col('price').cast('double'))
                   .withColumn('discountPercentage',col('discountPercentage').cast('double'))
                   .withColumn('stock',col('stock').cast('double'))
                   .withColumn('rating',col('rating').cast('double'))
                   .withColumn('ingestion_timestamps', current_timestamp()))
                   
```

```python
display(silver_customers)
display(silver_orders)
display(silver_products)
```

```python
silver_customers.write.format('delta').mode('overwrite').saveAsTable('silver_customers')
silver_orders.write.format('delta').mode('overwrite').saveAsTable('silver_orders')
silver_products.write.format('delta').mode('overwrite').saveAsTable('silver_products')
```

```python
from pyspark.sql.functions import col, trim, lower, current_timestamp

customers = spark.table("bronze_customers")

silver_customers = (
    customers
    .dropDuplicates(["id"])
    .withColumn("firstName", trim(col("firstName")))
    .withColumn("lastName", trim(col("lastName")))
    .withColumn("email", lower(trim(col("email"))))
    .withColumn("gender", lower(trim(col("gender"))))
    .withColumn("ingestion_timestamp", current_timestamp())
)

silver_customers.write.format("delta").mode("overwrite").saveAsTable(
    "silver_customers"
)
```

```python
from pyspark.sql.functions import col, sum, avg, count, round, explode

silver_orders = spark.table("silver_orders")

silver_orders.printSchema()
```

```python
sales_summary = (
    silver_orders
    .agg(
        count("id").alias("total_orders"),
        sum("total").alias("total_sales"),
        sum("discountedTotal").alias("total_discounted_sales"),
        sum("totalProducts").alias("total_products"),
        sum("totalQuantity").alias("total_quantity"),
        avg("total").alias("average_order_value")
    )
    .withColumn(
        "total_sales",
        round(col("total_sales"), 2)
    )
    .withColumn(
        "total_discounted_sales",
        round(col("total_discounted_sales"), 2)
    )
    .withColumn(
        "average_order_value",
        round(col("average_order_value"), 2)
    )
)

display(sales_summary)
```

```python
sales_summary.write.format('delta').mode('overwrite').saveAsTable('gold_sales_summary')
```

```python
order_products = silver_orders.select(
    col("id").alias("order_id"),
    col("userId").alias("customer_id"),
    explode(col("products")).alias("product")
)

display(order_products)
```

```python

silver_orders = spark.table("silver_orders")
silver_orders.printSchema()
```

```python
from pyspark.sql.functions import col

order_products = order_products.select(
    "order_id",
    "customer_id",
    col("product.id").alias("product_id"),
    col("product.title").alias("product_title"),
    col("product.quantity").alias("quantity"),
    col("product.price").alias("price"),
    col("product.total").alias("product_total"),
    col("product.discountedTotal").alias("discounted_total"),
    col("product.discountPercentage").alias("discount_percentage")
)

display(order_products)
```

```python
from pyspark.sql.functions import sum, count, round, col

product_performance = (
    order_products
    .groupBy("product_id", "product_title")
    .agg(
        sum("quantity").alias("total_quantity_sold"),
        sum("product_total").alias("gross_sales"),
        sum("discounted_total").alias("discounted_sales"),
        count("order_id").alias("order_count")
    )
    .withColumn("gross_sales", round(col("gross_sales"), 2))
    .withColumn("discounted_sales", round(col("discounted_sales"), 2))
)

display(product_performance)
```

```python
product_performance.write.format('delta').mode('overwrite').saveAsTable('gold_product_performance')
```

```python
from pyspark.sql.functions import col, sum, count, round

silver_orders = spark.table("silver_orders")

customer_summary = (
    silver_orders
    .groupBy(col("userId").alias("customer_id"))
    .agg(
        count("id").alias("total_orders"),
        sum("total").alias("total_order_value"),
        sum("discountedTotal").alias("total_discounted_value")
    )
    .withColumn("total_order_value", round(col("total_order_value"), 2))
    .withColumn("total_discounted_value", round(col("total_discounted_value"), 2))
)

display(customer_summary)
```

```python
customer_summary.write.format("delta").mode("overwrite").saveAsTable("gold_customer_summary")
display(spark.table("gold_customer_summary"))
```
